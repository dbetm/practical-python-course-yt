```python
features = ["temperature", "radius"]
target_name = "abs_mag"


with open('data/final_dataset.csv', 'r') as f:
    raw_records = f.readlines()


records = []

for idx, raw_record in enumerate(raw_records):
    if idx == 0:
        #ignore header
        continue

    temperature, radius, abs_mag = map(float, raw_record.replace("\n", "").split(","))

    records.append(
        {
            "temperature": temperature,
            "radius": radius,
            "abs_mag": abs_mag,
        }
    )

records
```

```python
from regression_tree import RegressionTree


model = RegressionTree(records, features, target_name)
model.train(depth=5)
```

```python
def mse_score(y_true: list, y_pred: list) -> float:
    n = len(y_true)
    return sum((true - pred)**2 for true, pred in zip(y_true, y_pred)) / n
```

```python
import random

y_gt = list()
y_pred = list()


for record in records:
    y_gt.append(record["abs_mag"])
    y_pred.append(model.predict(record))

    #print(examples[idx], model.predict(examples[idx]))


mse_score(y_gt, y_pred)
```
